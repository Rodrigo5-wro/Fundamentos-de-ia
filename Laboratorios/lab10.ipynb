```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
grafo=nx.DiGraph()
```

```python
grafo.add_weighted_edges_from([
    ("Norte", "Este", 8),
    ("Este", "Aeropuerto", 20),
    ("Oeste", "Aeropuerto", 18),
    ("Centro", "Norte", 10),
    ("Norte", "Sur", 30),
    ("Centro", "Sur", 15),
    ("Centro", "Este", 25),
    ("Norte", "Oeste", 15),
    ("Sur", "Oeste", 9)
])
```

```python
pos = nx.spring_layout(grafo, seed=10)

nx.draw(
    grafo,
    pos,
    with_labels=True,
    node_size=1500,
    arrows=True
)

labels = nx.get_edge_attributes(grafo, "weight")

nx.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels=labels
)
plt.show()
```

```python
# Mostramos las rutas
rutas = list(
    nx.all_simple_paths(
        grafo,
        source="Centro",
        target="Aeropuerto"
    )
)

# Imprimimos las rutas
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
# Calculamos el costo
def costo_ruta(grafo, ruta):
    costo = 0
    for i in range(len(ruta) - 1):
        costo += grafo[ruta[i]][ruta[i + 1]]["weight"]
    return costo

# Imprimimos el costo
for i, ruta in enumerate(rutas, start=1):
    costo = costo_ruta(grafo, ruta)
    print(f"Ruta {i}: {ruta} -> {costo} unidades")
```

```python
# Elegir la ruta más corta

costos = [
    costo_ruta(grafo, ruta)
    for ruta in rutas
]

mejor_ruta = costos.index(min(costos))
```

```python
# Imprimir la mejor ruta

print(f"La mejor ruta es: {mejor_ruta + 1}")
print(f"El camino costo es: {costos[mejor_ruta]}")
```

```python
ruta_mas_corta = nx.dijkstra_path(
    grafo,
    source="Centro",
    target="Aeropuerto",
    weight="weight"
)

tiempo_total = nx.dijkstra_path_length(
    grafo,
    source="Centro",
    target="Aeropuerto",
    weight="weight"
)

# Imprimir

print(f"La ruta más corta es: {ruta_mas_corta}")
print(f"El tiempo total es: {tiempo_total} minutos")
```
