```python
#librería utilizada para graficar
import matplotlib.pyplot as plt
#librería que me permite trabajar con estructuras complejas, como Grafos
import networkx as nx
```

```python
#Creamos una instancia de la clase Graph
#Graph es una colección de nodos
grafoBeach = nx.Graph()
```

```python
grafoBeach.add_edges_from(
    [
        ("Servidor_1", "RouterA"),
        ("RouterA", "SwitchA"),
        ("SwitchA", "PC1"),
        ("SwitchA", "ImpresoraA"),
        ("RouterA", "SwitchB"),
        ("SwitchB", "PC2"),
        ("SwitchB", "ImpresoraA")
    ]
)
```

```python
#Mostrar el grafo
pos = nx.spring_layout(grafoBeach)

nx.draw(
    grafoBeach,
    pos,
    with_labels=True,
    node_size=2000
)

plt.show()
```
