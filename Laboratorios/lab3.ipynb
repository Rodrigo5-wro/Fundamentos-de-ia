Laboratorio 3: Proposiciones
### Objetivo: Un VideoJuego uitliza un agente inteligente enemigo que debe decidier si puede iniciar un ataque contra el jugador.

#Proposiciones
P: Se encuentra dentro del rango
Q: El nivel de vida es suficiente
R: Existe municiones disponibles
S: Este visible
T: No exista amenaza cercana
U: No esta aturdido

```python
# Difinicion de variables
rango = True #Si esta dentro del rango
nivelVida = True #Si tiene vida suficiente
municiones = False # Si cuenta con municiones disponibles
visible = True # si esta visible el jugador 
amenaza = False # No existe amenaza cercana
aturdido = False # No esta aturdido
```

```python
puedeAtacar = (rango==True and visible==True and nivelVida==True and municiones==True and amenaza==False and aturdido==False)
```

```python
print("El agente enemigo puede atacar?", puedeAtacar)
```

```python
print ("== Angente Enemigo Inteligente ==")
print ("Development by rodrigo")

rango = input("El jugador se encuentra dentro del rango(s/n): ").lower() == "s"
visible = input("El jugador esta visible(s/n): ").lower() == "s"
nivelvida = input("El enemigo tiene suficiente vida(s/n): ").lower() == "s"
municiones = input("El enemigo tiene suficiente municiones(s/n): ").lower() == "s"
amenaza = input("Existe una amenaza cercana?(s/n): ").lower() == "n"
aturdido = input("El enemigo no esta aturdido?(s/n): ").lower() == "n"

puedeAtacar = (rango==True and visible==True and nivelVida==True and municiones==True and amenaza==True and aturdido==True)

print ("Decision del agente")
if puedeAtacar == True:
    print ("El agente puede atacar: es tu momento")
else:
    print("Escondete No es momento de atacar")
```
