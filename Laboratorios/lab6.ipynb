```python
import matplotlib.pyplot as plt
import networkx as nx
```

```python
grafo = nx.Graph()
```

```python
grafo.add_edges_from(
    [
        ("A","B"),
        ("A","C"),
        ("B","D"),
        ("C","G"),
        ("D","F"),
        ("G","E"),
        ("G","J"),
        ("E","F"),
        ("F","H"),
        ("H","I"),
        ("I","J")
    ]
)
```

```python
# Mostrar el Grafo
nx.draw(
    grafo,
    with_labels =True,
    node_size =1000
)
plt.show()
```

```python
rutas = list(
    nx.all_simple_paths(grafo, source="A", target="J")
)
```

```python
#Imprimir las rutas
print("Rutas encontradas:")
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
#Mostrar el total de pasos de cada ruta 
pasos1 = len(rutas[0])-1
print("Pasos de la ruta 1: ", pasos1)
pasos2 = len(rutas[1])-1
print("Pasos de la ruta 2: ", pasos2)
pasos3 = len(rutas[2])-1
print("Pasos de la ruta 3: ", pasos3)
pasos4 = len(rutas[3])-1
print("Pasos de la ruta 4: ", pasos4)
```

```python
# Aplicamos la logica proposicional
P = pasos1 < pasos2 #P la mejor ruta es la 1
Q = pasos2 < pasos1 # Q: La mejor ruta es la 2
R = pasos3 < pasos4
S = pasos4 < pasos3

if P:
    print("P = True")
    print("La ruta 1 es mejor opción.")
elif Q:
    print("Q = True")
    print("La ruta 2 es mejor opción")
elif R:
    print("R = True")
    print("La ruta 3 es mejor opcion")
elif S:
    print("S = True")
    print("La ruta 4 es mjor ")
else:
    print("P= False")
    print("Q  = False")
    print("R = False")
    print("S = False")
    print("Las dos rutas son igual de buenas, haz lo que quieras")
```
